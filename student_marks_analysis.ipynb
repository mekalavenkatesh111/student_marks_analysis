```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#step 1:create the data
data = {
    "Name":["Harish","venkay","rajesh","syam","venkataramana","Hemanth"],
    "Marks":[12,32,44,55,66,66],
    "Age":[18,19,20,21,22,20],
    "Maths":[66,79,78,42,55,99],
    "Social":[33,55,34,45,66,77],
     i   "English":[33,55,44,55,77,None]

}
df = pd.DataFrame(data)
print(df)

#step 2:Look at the data
print(df.head())
print(df.tail())
print(df.describe())

#step 3:Check and fill missing values
print(df.isnull().sum())
df["English"] = df["English"].fillna(df["English"].mean())
print(df)

#step 4:Answer questions
print(df.loc[df["Maths"].idxmax(),"Name"])
failed =df[df["Maths"] < 40]
print(failed)

#step 5:Total and Rank
df["Total"] = df["Maths"] + df["Social"] + df["English"]
df["Rank"] = df["Total"].rank(ascending=False).astype(int)
print(df.sort_values("Rank"))

#step 6:Bar chart
df[["Maths","Social","English"]].mean().plot(kind="bar")
plt.title("Average Marks by Subject")
plt.ylabel("Average Marks")
plt.show()
```
